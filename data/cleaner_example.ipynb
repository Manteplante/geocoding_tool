# Cleaner example — Norwegian towns

Small, self-contained demo of the pattern in `data/cleaner.ipynb`: read messy
input, clean it, write it back out. Run via `make clean-notebook`.

Input: `data/input/example.csv` — five Norwegian towns (Otta, Vinstra,
Lørenskog, Stryn, Bodø) with deliberately messy formatting: stray whitespace,
inconsistent casing, a trailing "kommune", and mixed-case country names.

## Read data from input folder

```python
import pandas as pd

from geocoding_tool import project_root

ROOT = project_root()
INPUT_PATH = ROOT / "data" / "input" / "example.csv"
OUTPUT_PATH = ROOT / "data" / "output" / "example_clean.csv"

df = pd.read_csv(INPUT_PATH, dtype="string")
df
```

## Clean data

- trim whitespace
- drop a trailing "kommune" suffix
- title-case the town and country names
- drop exact duplicate rows

```python
cleaned = df.copy()
cleaned["raw_town"] = cleaned["raw_town"].str.strip()
cleaned["raw_town"] = cleaned["raw_town"].str.replace(
    r"\s+kommune$", "", regex=True, case=False
)

cleaned["town"] = cleaned["raw_town"].str.title()
cleaned["country"] = cleaned["country"].str.strip().str.title()

cleaned = cleaned.drop(columns=["raw_town"]).drop_duplicates().reset_index(drop=True)
cleaned
```

## Export output to output folder

```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
cleaned.to_csv(OUTPUT_PATH, index=False)

print(f"wrote {len(cleaned)} rows to {OUTPUT_PATH.relative_to(ROOT)}")
cleaned
```
